# Lancer l’interface de validation

Ce notebook lance le serveur local qui affiche les PDF et les annotations. Le fichier `validation_web/serveur.py` reste nécessaire car il sert les pages au navigateur.

```python
from pathlib import Path
import subprocess
import sys

def trouver_racine() -> Path:
    for candidat in (Path.cwd().resolve(), *Path.cwd().resolve().parents):
        if (candidat / "donnees" / "comptes_rendus" / "total").is_dir():
            return candidat
    raise FileNotFoundError("Corpus donnees/comptes_rendus/total introuvable")


RACINE = trouver_racine()
SERVEUR = RACINE / "extraction_evaluation" / "validation_web" / "serveur.py"
DOSSIER_VALIDATION = RACINE / "extraction_evaluation" / "evaluation" / "lots"
LOT = "lots/evaluation_30.csv"
PORT = 8766
```

```python
serveur = subprocess.Popen([
    sys.executable, str(SERVEUR),
    "--dossier-validation", str(DOSSIER_VALIDATION),
    "--lot", LOT,
    "--port", str(PORT),
])
print(f"Interface : http://localhost:{PORT}")
print("Pour arrêter : serveur.terminate()")
```
